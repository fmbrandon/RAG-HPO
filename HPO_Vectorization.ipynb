# Reproducible HPO vectorization

Vector generation is implemented by the installed `rag_hpo` package. This notebook is intentionally network-free when executed.

````python
from rag_hpo.embeddings import DEFAULT_SAPBERT_MODEL, DEFAULT_SAPBERT_REVISION
from rag_hpo.ontology import DEFAULT_HPO_URL

configuration = {
    "hpo_source": DEFAULT_HPO_URL,
    "embedding_model": DEFAULT_SAPBERT_MODEL,
    "embedding_revision": DEFAULT_SAPBERT_REVISION,
}
print(configuration)
````

Run a bounded build from a terminal:

```bash
rag-hpo vectorize --hpo-addons HPO_addons.csv --output-dir artifacts/hpo --limit 50
```

Remove `--limit 50` only when intentionally generating the full release artifact.
